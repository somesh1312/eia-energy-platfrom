```python
import requests
import pandas as pd

url = "https://api.eia.gov/v2/seds/data/"

all_rows = []
offset = 0
page_size = 5000

while True:

    params = {
        "api_key": API_KEY,
        "frequency": "annual",
        "data[0]": "value",
        "start": "2014",
        "end": "2024",
        "sort[0][column]": "stateId",
        "sort[0][direction]": "desc",
        "offset": offset,
        "length": page_size
    }

    response = requests.get(url, params=params)
    response.raise_for_status()

    result = response.json()["response"]

    rows = result["data"]

    if not rows:
        break

    all_rows.extend(rows)

    print(f"Downloaded {len(all_rows)} / {result['total']}")

    offset += page_size

    if len(all_rows) >= int(result["total"]):
        break


df = pd.DataFrame(all_rows)

print(df.shape)
```
